```python
def normalisieren(name):
    name = name.lower()
    name = name.replace("saint-", "st-").replace("sainte-", "ste-")
    name = name.replace("-", " ").replace("/", " ")
    return name.strip()

ortschaften["name_norm"] = ortschaften["Ortschaftsname"].apply(normalisieren)
```

```python
def kanton_finden(name):
    n = normalisieren(name)

    # 1. Versuch: exakter Treffer
    treffer = ortschaften[ortschaften["name_norm"] == n]

    # 2. Versuch: Ortschaft beginnt mit dem Namen (z. B. "davos" -> "davos glaris")
    if treffer.empty:
        treffer = ortschaften[ortschaften["name_norm"].str.startswith(n + " ")]

    kantone = treffer["Kantonskürzel"].unique()

    if len(kantone) == 1:
        return kantone[0]
    elif len(kantone) > 1:
        return "mehrdeutig"
    else:
        return None
```

```python
result["start_kanton"] = result["start_ort"].apply(kanton_finden)
result["end_kanton"] = result["end_ort"].apply(kanton_finden)

for spalte in ["start_kanton", "end_kanton"]:
    werte = result[spalte].fillna("nicht gefunden")
    print(spalte)
    print("  eindeutig:      ", (~werte.isin(["nicht gefunden", "mehrdeutig"])).mean().round(3))
    print("  mehrdeutig:     ", (werte == "mehrdeutig").mean().round(3))
    print("  nicht gefunden: ", (werte == "nicht gefunden").mean().round(3))
```

```python
def kanton_finden(name):
    n = normalisieren(name)

    treffer = ortschaften[ortschaften["name_norm"] == n]
    if treffer.empty:
        treffer = ortschaften[ortschaften["name_norm"].str.startswith(n + " ")]

    kantone = treffer["Kantonskürzel"].unique()

    # Regel A: bei mehreren Kantonen die gleichnamige Gemeinde bevorzugen
    if len(kantone) > 1:
        gleichnamig = treffer[treffer["Gemeindename"].apply(normalisieren) == n]
        kantone = gleichnamig["Kantonskürzel"].unique() if not gleichnamig.empty else kantone

    if len(kantone) == 1:
        return kantone[0]
    elif len(kantone) > 1:
        return "mehrdeutig"
    else:
        return None
```

```python
def kandidaten(name):
    n = normalisieren(name)
    treffer = ortschaften[ortschaften["name_norm"] == n]
    if treffer.empty:
        treffer = ortschaften[ortschaften["name_norm"].str.startswith(n + " ")]
    return set(treffer["Kantonskürzel"])

def aufloesen(row, eigene, andere, ort):
    if row[eigene] == "mehrdeutig" and row[andere] in kandidaten(row[ort]):
        return row[andere]
    return row[eigene]

result["start_kanton"] = result.apply(aufloesen, axis=1, args=("start_kanton", "end_kanton", "start_ort"))
result["end_kanton"] = result.apply(aufloesen, axis=1, args=("end_kanton", "start_kanton", "end_ort"))
```

```python
import re

AUSLAND = {"Balzers", "Gaflei", "Malbun", "Schellenberg", "Triesenberg", "Mont Salève"}
ZUSAETZE = ["Mittelstation", "Bergstation", "Station", "Gare", "Berghaus"]

def kanton_finden_erweitert(name):
    # Gruppe 1: Ausland
    if name in AUSLAND:
        return "Ausland"

    # Gruppe 2: Zusätze wie "Bergstation" oder "Gare" entfernen
    for wort in ZUSAETZE:
        name = name.replace(wort, "")
    name = name.strip()
    kanton = kanton_finden(name)
    if kanton is not None:
        return kanton

    # Gruppe 3a: nur den Teil vor Bindestrich oder " bei " nehmen
    teil = re.split(r"-| bei ", name)[0].strip()
    kanton = kanton_finden(teil)
    if kanton is not None:
        return kanton

    # Gruppe 3b: Name als ganzes Wort irgendwo im Ortschaftsnamen (z. B. "Glaris" in "Davos Glaris")
    n = normalisieren(name)
    treffer = ortschaften[ortschaften["name_norm"].str.contains(rf"\b{re.escape(n)}\b", regex=True)]
    kantone = treffer["Kantonskürzel"].unique()
    if len(kantone) == 1:
        return kantone[0]

    return None
```

```python
import re
import time

def kanton_swisstopo(name):
    params = {"searchText": name, "type": "locations", "origins": "gazetteer", "limit": 20}
    try:
        ergebnisse = requests.get(url, params=params, timeout=10).json()["results"]
    except Exception:
        return None

    kantone = set()
    for e in ergebnisse:
        label = e["attrs"]["label"]
        fett = re.search(r"<b>(.*?)</b>", label)
        kanton = re.search(r"\(([A-Z]{2})\)", label)
        if fett and kanton and normalisieren(fett.group(1)) == normalisieren(name):
            kantone.add(kanton.group(1))

    if len(kantone) == 1:
        return kantone.pop()
    elif len(kantone) > 1:
        return "mehrdeutig"
    return None
```

```python
noch_fehlend = pd.concat([
    result.loc[result["start_kanton"].isna(), "start_ort"],
    result.loc[result["end_kanton"].isna(), "end_ort"]
]).unique()

for name in sorted(noch_fehlend):
    print(name, "→", kanton_swisstopo(name))
    time.sleep(0.5)
```

```python
swisstopo_ergebnis = {}
for name in noch_fehlend:
    swisstopo_ergebnis[name] = kanton_swisstopo(name)
    time.sleep(0.5)

for ort, kanton in [("start_ort", "start_kanton"), ("end_ort", "end_kanton")]:
    maske = result[kanton].isna()
    result.loc[maske, kanton] = result.loc[maske, ort].map(swisstopo_ergebnis)
```
