```python
import re

AUSLAND = {"Balzers", "Gaflei", "Malbun", "Schellenberg", "Triesenberg", "Mont Salève"}
ZUSAETZE = ["Mittelstation", "Bergstation", "Station", "Gare", "Berghaus"]

def kanton_finden_erweitert(name):
    # Gruppe 1: Ausland
    if name in AUSLAND:
        return "Ausland"

    # Gruppe 2: Zusätze wie "Bergstation" oder "Gare" entfernen
    for wort in ZUSAETZE:
        name = name.replace(wort, "")
    name = name.strip()
    kanton = kanton_finden(name)
    if kanton is not None:
        return kanton

    # Gruppe 3a: nur den Teil vor Bindestrich oder " bei " nehmen
    teil = re.split(r"-| bei ", name)[0].strip()
    kanton = kanton_finden(teil)
    if kanton is not None:
        return kanton

    # Gruppe 3b: Name als ganzes Wort irgendwo im Ortschaftsnamen (z. B. "Glaris" in "Davos Glaris")
    n = normalisieren(name)
    treffer = ortschaften[ortschaften["name_norm"].str.contains(rf"\b{re.escape(n)}\b", regex=True)]
    kantone = treffer["Kantonskürzel"].unique()
    if len(kantone) == 1:
        return kantone[0]

    return None
```

```python
import re
import time

def kanton_swisstopo(name):
    params = {"searchText": name, "type": "locations", "origins": "gazetteer", "limit": 20}
    try:
        ergebnisse = requests.get(url, params=params, timeout=10).json()["results"]
    except Exception:
        return None

    kantone = set()
    for e in ergebnisse:
        label = e["attrs"]["label"]
        fett = re.search(r"<b>(.*?)</b>", label)
        kanton = re.search(r"\(([A-Z]{2})\)", label)
        if fett and kanton and normalisieren(fett.group(1)) == normalisieren(name):
            kantone.add(kanton.group(1))

    if len(kantone) == 1:
        return kantone.pop()
    elif len(kantone) > 1:
        return "mehrdeutig"
    return None
```

```python
STATIONSWOERTER = ["Mittelstation", "Bergstation", "Talstation", "Station", "Berghaus"]

def suchvarianten(name):
    varianten = [name]
    bereinigt = name
    for wort in STATIONSWOERTER:
        bereinigt = bereinigt.replace(wort, "")
    bereinigt = bereinigt.strip()
    if bereinigt != name:
        varianten.append(bereinigt)
    if bereinigt.endswith("bahn"):
        varianten.append(bereinigt[:-4])  # "Sidelhornbahn" -> "Sidelhorn"
    return varianten

def kanton_swisstopo_varianten(name):
    for variante in suchvarianten(name):
        ergebnis = kanton_swisstopo(variante)
        time.sleep(0.5)
        if ergebnis is not None:
            return ergebnis
    return None
```

```python
def normalisieren(name):
    name = name.lower()
    name = name.replace("’", "'").replace("‘", "'")   # Apostrophe vereinheitlichen
    name = name.replace("saint-", "st-").replace("sainte-", "ste-")
    name = name.replace("-", " ").replace("/", " ")
    return name.strip()

def kanton_swisstopo(name):
    params = {"searchText": name, "type": "locations", "origins": "gazetteer", "limit": 20}
    try:
        ergebnisse = requests.get(url, params=params, timeout=10).json()["results"]
    except Exception:
        return None

    n = normalisieren(name)
    streng, locker = set(), set()

    for e in ergebnisse:
        label = e["attrs"]["label"]
        fett = re.search(r"<b>(.*?)</b>", label)
        kanton = re.search(r"\(([A-Z]{2})\)", label)
        if not (fett and kanton):
            continue
        fett_n = normalisieren(fett.group(1))
        gemeinde_n = normalisieren(label.split(" - ")[-1])

        if fett_n == n:
            streng.add(kanton.group(1))
        elif fett_n.startswith(n + " ") or gemeinde_n == n:
            locker.add(kanton.group(1))

    kantone = streng if streng else locker
    if len(kantone) == 1:
        return kantone.pop()
    elif len(kantone) > 1:
        return "mehrdeutig"
    return None
```

```python
def kantone_swisstopo(name):
    params = {"searchText": name, "type": "locations", "origins": "gazetteer", "limit": 20}
    try:
        ergebnisse = requests.get(url, params=params, timeout=10).json()["results"]
    except Exception:
        return set()
    n = normalisieren(name)
    streng = set()
    for e in ergebnisse:
        label = e["attrs"]["label"]
        fett = re.search(r"<b>(.*?)</b>", label)
        kanton = re.search(r"\(([A-Z]{2})\)", label)
        if fett and kanton and normalisieren(fett.group(1)) == n:
            streng.add(kanton.group(1))
    return streng

# Kandidaten für alle mehrdeutigen Namen sammeln (aus beiden Quellen)
mehrdeutige_namen = pd.concat([
    result.loc[result["start_kanton"] == "mehrdeutig", "start_ort"],
    result.loc[result["end_kanton"] == "mehrdeutig", "end_ort"]
]).unique()

kandidaten_alle = {}
for name in mehrdeutige_namen:
    kandidaten_alle[name] = kandidaten(name) | kantone_swisstopo(name)
    time.sleep(0.5)

for name, k in kandidaten_alle.items():
    print(name, "→", sorted(k))
```

```python
def kanton_mit_fallback(name):
    # Regel 1: "X bei Y" -> zuerst X, dann Y versuchen
    teile = [t.strip() for t in name.split(" bei ")]
    for teil in teile:
        # Regel 2: Name schrittweise um ein Wort kürzen
        woerter = teil.split()
        while woerter:
            versuch = " ".join(woerter)
            ergebnis = kanton_finden(versuch)          # Ortschaftenverzeichnis
            if ergebnis is None:
                ergebnis = kanton_swisstopo(versuch)   # swisstopo
                time.sleep(0.5)
            if ergebnis is not None and ergebnis != "mehrdeutig":
                return ergebnis
            woerter = woerter[:-1]
    return None
```

```python
ALLGEMEINE_WOERTER = {"alp", "alpe", "col", "monte", "piz", "pass", "le", "la", "les", "il", "di", "san", "st"}

def kanton_mit_fallback(name, zeigen=False):
    teile = [t.strip() for t in name.split(" bei ")]
    for teil in teile:
        woerter = teil.split()
        while woerter:
            versuch = " ".join(woerter)
            if normalisieren(versuch) in ALLGEMEINE_WOERTER:
                break                      # nicht auf ein allgemeines Wort kürzen
            ergebnis = kanton_finden(versuch)
            if ergebnis is None:
                ergebnis = kanton_swisstopo(versuch)
                time.sleep(0.5)
            if ergebnis is not None and ergebnis != "mehrdeutig":
                if zeigen:
                    print(f"   gefunden als: '{versuch}'")
                return ergebnis
            woerter = woerter[:-1]
    return None

for name in ["Alp Camp", "Cantine di Gandria", "Sils i. E.", "Sommerau bei Rümlingen"]:
    print(name, "→", kanton_mit_fallback(name, zeigen=True))
```

```python
fallback_ergebnis = {}
for name in ["Alp Camp", "Cantine di Gandria", "Sils i. E.", "Sommerau bei Rümlingen"]:
    fallback_ergebnis[name] = kanton_mit_fallback(name)

for ort, kanton in [("start_ort", "start_kanton"), ("end_ort", "end_kanton")]:
    maske = result[kanton].isna()
    result.loc[maske, kanton] = result.loc[maske, ort].map(fallback_ergebnis)

stand()
```

```python
TITEL_STOPPWOERTER = {
    "weg", "panoramaweg", "rundweg", "höhenweg", "wanderweg", "trail", "tour", "via",
    "sentier", "circuit", "chemin", "sentiero", "giro", "percorso",
    "de", "du", "des", "la", "le", "les", "di", "del", "della", "im", "am", "und", "et",
    "der", "die", "das", "von", "zum", "zur", "ob", "sur", "en", "e"
}

def titel_phrasen(titel):
    woerter = [w for w in re.split(r"[\s\-–]+", titel)
               if w and normalisieren(w) not in TITEL_STOPPWOERTER and not w.isdigit()]
    phrasen = []
    for laenge in [3, 2, 1]:                       # längere Wortgruppen zuerst
        for i in range(len(woerter) - laenge + 1):
            phrasen.append(" ".join(woerter[i:i + laenge]))
    return phrasen

def kanton_aus_titel(titel, kandidaten_menge, zeigen=False):
    for phrase in titel_phrasen(titel):
        gefunden = kandidaten(phrase) | kantone_swisstopo(phrase)
        time.sleep(0.5)
        passend = gefunden & kandidaten_menge
        if len(passend) == 1:
            if zeigen:
                print(f"   gefunden über: '{phrase}'")
            return passend.pop()
    return None
```

```python
GIPFEL_TYPEN = {"alpiner gipfel", "hauptgipfel", "gipfel", "haupthuegel", "huegel",
                "pass", "strassenpass", "aussichtspunkt"}

def kanton_gipfel(name, zeigen=False):
    params = {"searchText": name, "type": "locations", "origins": "gazetteer", "limit": 50}
    try:
        ergebnisse = requests.get(url, params=params, timeout=10).json()["results"]
    except Exception:
        return None
    n = normalisieren(name)
    kantone = set()
    for e in ergebnisse:
        label = e["attrs"]["label"]
        typ = re.search(r"<i>(.*?)</i>", label)
        fett = re.search(r"<b>(.*?)</b>", label)
        k = re.search(r"\(([A-Z]{2}(?:,[A-Z]{2})*)\)", label)
        if typ and fett and k and normalisieren(fett.group(1)) == n \
                and typ.group(1).lower() in GIPFEL_TYPEN:
            kantone.update(k.group(1).split(","))
            if zeigen:
                print("   ", label)
    return kantone.pop() if len(kantone) == 1 else None
```

```python
GIPFEL_TYPEN = {"alpiner gipfel", "hauptgipfel", "gipfel", "haupthuegel", "huegel",
                "pass", "strassenpass"}          # ohne "aussichtspunkt"

def ist_eindeutig(k):
    return isinstance(k, str) and k not in ("mehrdeutig", "Ausland")

df_neu = result.copy()
methode = {}
paare = [("start_kanton", "start_ort", "end_kanton"), ("end_kanton", "end_ort", "start_kanton")]

# 1. Regel B: anderer Endpunkt
for idx, row in df_neu.iterrows():
    for eigen, ort, anders in paare:
        if df_neu.at[idx, eigen] == "mehrdeutig" and ist_eindeutig(df_neu.at[idx, anders]) \
                and df_neu.at[idx, anders] in kandidaten_alle.get(row[ort], set()):
            df_neu.at[idx, eigen] = df_neu.at[idx, anders]
            methode[(idx, eigen)] = "Regel B"

# 2. Gipfelregel: Lage laut swisstopo
gipfel_cache = {}
for idx, row in df_neu.iterrows():
    for eigen, ort, _ in paare:
        if df_neu.at[idx, eigen] == "mehrdeutig":
            name = row[ort]
            if name not in gipfel_cache:
                suchname = "Wasserfallen" if name == "Wasserfallen Bergstation" else name
                gipfel_cache[name] = kanton_gipfel(suchname)
                time.sleep(0.5)
            if gipfel_cache[name]:
                df_neu.at[idx, eigen] = gipfel_cache[name]
                methode[(idx, eigen)] = "Gipfel"

# 3. Titelregel
for idx, row in df_neu.iterrows():
    for eigen, ort, _ in paare:
        if df_neu.at[idx, eigen] == "mehrdeutig":
            k = kanton_aus_titel(row["title"], kandidaten_alle.get(row[ort], set()))
            if k:
                df_neu.at[idx, eigen] = k
                methode[(idx, eigen)] = "Titel"

# Übersicht der Änderungen
for (idx, eigen), m in methode.items():
    ort = "start_ort" if eigen == "start_kanton" else "end_ort"
    print(f"{df_neu.at[idx, 'title']} | {df_neu.at[idx, ort]} → {df_neu.at[idx, eigen]} ({m})")
```
