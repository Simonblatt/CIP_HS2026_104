```python
# This notebook answers RQ1: Which cantons have the highest density of outdoor routes with a maximum altitude below 1200 m (accessible in the shoulder seasons)?

import pandas as pd
# pandas: works with tables (DataFrames).
# Used to load the route tables, filter them and count routes per canton.

#openpyxl ist required for reading excel-files
%pip install openpyxl

```

```python
# The four route tables with cantons (result of canton_assignment.ipynb) are stored in a list.
# The four route tables with cantons (result of canton_assignment.ipynb)
files = [
    "result_wanderland_cantons.csv",
    "result_veloland_cantons.csv",
    "result_mountainbikeland_cantons.csv",
    "result_skatingland_cantons.csv",
]
```

```python
# Load the four files and combine them into one table with all routes
routes = pd.concat([pd.read_csv(file) for file in files], ignore_index=True)
```

```python
# Check: number of routes per activity and in total
print(routes["land"].value_counts())
print("Total:", len(routes))
```

```python
# Keep only routes whose highest point is below 1200 m
# (assumed to be accessible in spring and autumn, see RQ1)
low_routes = routes[routes["el_max_m"] < 1200]

# Check: number of routes before and after the filter
print("All routes:", len(routes))
print("Routes below 1200 m:", len(low_routes))
print("Routes without altitude information:", routes["el_max_m"].isna().sum())
```

```python
low_routes
```

```python
low_routes.to_csv("rq1_routes_below_1200m.csv", index=False)
```

```python
# Count the routes per canton: a route counts for every canton in which it starts or ends.
# A route that starts and ends in the same canton counts only once for that canton.

# 1. One row per route and canton: put start cantons and end cantons below each other
route_canton = pd.concat([
    low_routes[["start_canton"]].rename(columns={"start_canton": "canton"}),
    low_routes[["end_canton"]].rename(columns={"end_canton": "canton"}),
])

# 2. Remove duplicates: same route (same row number) with the same canton counts once
route_canton = route_canton.reset_index().drop_duplicates()

# 3. Routes abroad are not counted
route_canton = route_canton[route_canton["canton"] != "abroad"]

# 4. Count the routes per canton
routes_per_canton = route_canton["canton"].value_counts()
print(routes_per_canton.to_string())
```

```python
# Load the geographic key figures of the cantons (BFS, as of 01.01.2026)
# Contains for each canton: canton number (KTNR), name, area in hectares (AREA_HA) and elevation
canton_figures = pd.read_excel("Geographische_Kennzahlen.xlsx", sheet_name="Daten")

# Check: first rows
canton_figures[["KTNR", "KTNAME", "AREA_HA"]].head()
```

```python
# Canton abbreviation for each official canton number (order of the Federal Constitution)
canton_abbreviations = {
    1: "ZH", 2: "BE", 3: "LU", 4: "UR", 5: "SZ", 6: "OW", 7: "NW", 8: "GL", 9: "ZG",
    10: "FR", 11: "SO", 12: "BS", 13: "BL", 14: "SH", 15: "AR", 16: "AI", 17: "SG",
    18: "GR", 19: "AG", 20: "TG", 21: "TI", 22: "VD", 23: "VS", 24: "NE", 25: "GE", 26: "JU"
}
canton_figures["canton"] = canton_figures["KTNR"].map(canton_abbreviations)

# Convert the area from hectares to square kilometres (1 km² = 100 ha)
canton_figures["area_km2"] = canton_figures["AREA_HA"] / 100

# Check: number, abbreviation, name and area
canton_figures[["KTNR", "canton", "KTNAME", "area_km2"]]
```

```python
# Combine the number of routes with the area of each canton.
# The canton table is the base (how="left"), so all 26 cantons are included,
# also those without any route below 1200 m (e.g. Zug); their count is set to 0.
rq1 = canton_figures[["canton", "KTNAME", "area_km2"]].merge(
    routes_per_canton.rename("routes").reset_index(),
    on="canton",
    how="left"
)
rq1["routes"] = rq1["routes"].fillna(0).astype(int)

# Density: number of routes per 100 km² of canton area
rq1["density_per_100km2"] = (rq1["routes"] / rq1["area_km2"] * 100).round(2)

# Sort from highest to lowest density
rq1 = rq1.sort_values("density_per_100km2", ascending=False).reset_index(drop=True)
print(rq1.to_string())
```

```python
# Save the result of RQ1: routes below 1200 m and density per canton
rq1.to_csv("rq1_result_density_below_1200m.csv", index=False)
print("Saved:", len(rq1), "cantons")
```
